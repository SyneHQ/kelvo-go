```python
taxi = lab.taxi()
config = lab.catalog(sources=[{"id": "trips", "type": "parquet", "path": str(taxi)}])
quality = lab.query(config, """
    SELECT COUNT(*)::BIGINT AS total,
           COUNT(*) FILTER (WHERE tpep_pickup_datetime IS NULL)::BIGINT AS missing_pickup,
           COUNT(*) FILTER (WHERE passenger_count IS NULL)::BIGINT AS missing_passengers,
           COUNT(*) FILTER (WHERE trip_distance <= 0)::BIGINT AS nonpositive_distance,
           COUNT(*) FILTER (WHERE total_amount < 0)::BIGINT AS negative_total,
           COUNT(*) FILTER (WHERE tpep_pickup_datetime < TIMESTAMP '2024-01-01'
                             OR tpep_pickup_datetime >= TIMESTAMP '2024-02-01')::BIGINT AS outside_month
    FROM trips
""", sources=["trips"]).table().to_pylist()[0]
assert quality["total"] > 0
assert all(0 <= value <= quality["total"] for name, value in quality.items() if name != "total")
import pandas as pd
display(pd.DataFrame([{"check": key, "rows": value, "share_pct": value / quality["total"] * 100}
                      for key, value in quality.items() if key != "total"]))
```

## Define a report contract, not a claim that the data is perfect

Diagnostic counts can overlap. For reconciliation, classify every row exactly
once as eligible or excluded. NULL predicates enter the excluded population.


```python
eligibility = """
    tpep_pickup_datetime >= TIMESTAMP '2024-01-01'
    AND tpep_pickup_datetime < TIMESTAMP '2024-02-01'
    AND trip_distance > 0 AND trip_distance < 100
    AND total_amount >= 0 AND total_amount < 1000
"""
contract = lab.query(config, f"""
    WITH classified AS (
        SELECT CASE WHEN {eligibility} THEN 'eligible' ELSE 'excluded' END AS status
        FROM trips
    )
    SELECT status, COUNT(*)::BIGINT AS rows FROM classified GROUP BY status ORDER BY status
""", sources=["trips"]).table()
counts = dict(zip(contract["status"].to_pylist(), contract["rows"].to_pylist()))
assert sum(counts.values()) == quality["total"]
assert counts.get("eligible", 0) > 0
direct = lab.query(config, f"SELECT COUNT(*)::BIGINT AS n FROM trips WHERE {eligibility}",
                   sources=["trips"]).table()["n"][0].as_py()
assert direct == counts["eligible"]
display(contract.to_pandas())
```

## Review thresholds with a data owner

The threshold below is an example alert, not a guarantee of source quality.
Keep unexpected records available for investigation and version the agreed
filter whenever a dashboard's definition changes.


```python
REVIEW_THRESHOLD = 0.05
excluded_share = counts.get("excluded", 0) / quality["total"]
print({"excluded_share": excluded_share,
       "review_threshold": REVIEW_THRESHOLD,
       "review_required": excluded_share > REVIEW_THRESHOLD})
```
